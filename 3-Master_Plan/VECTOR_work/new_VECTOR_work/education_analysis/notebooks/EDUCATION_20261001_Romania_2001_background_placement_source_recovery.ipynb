```python
# 1. SETTINGS — these are the controls you may change before each run.
# Keep this first run on NT (Neamț) so you can see one county through to its source gate.
# Set COUNTIES = [] later to use the default small-county-first queue.
COUNTIES = []

# Maximum NEW archived pages and elapsed hours in one run. Verified cached pages
# do not count against MAX_NEW_PAGES. Neither limit writes missing data off.
MAX_NEW_PAGES = 800
MAX_HOURS = 2

# Random request spacing, in seconds. The worker also stops on HTTP 429.
MIN_INTERVAL_SECONDS = 31
MAX_INTERVAL_SECONDS = 55

# A prior HTTP 404 is kept unresolved and skipped by default. Change this only
# when YOU deliberately want to retry those same historical addresses.
RETRY_PREVIOUS_HTTP_404 = False

# Safety switch. Keep False while inspecting the queue; set True to fetch pages.
RUN_SOURCE_ACQUISITION = True

# Response 429 delay, in seconds.
DELAY_429 = 600

print("Settings loaded. Counties:", COUNTIES or "all remaining, small first")
print("New-page limit:", MAX_NEW_PAGES, "| Time limit (hours):", MAX_HOURS)
print("Network acquisition enabled:", RUN_SOURCE_ACQUISITION)
```

```python
# 2. Load the existing, reviewed Python worker. This notebook has no second
# copy of the retrieval logic; changes to that code are picked up on reload.
from pathlib import Path
import importlib
import sys

worker_filename = "EDUCATION_20261001_romania_background_placement_worker.py"
relative_code = Path("3-Master_Plan/VECTOR_work/new_VECTOR_work/education_analysis/code")
search_roots = [Path.cwd(), *Path.cwd().parents]
candidate_dirs = [root / relative_code for root in search_roots]
candidate_dirs += [root / "code" for root in search_roots]
CODE_DIR = next((folder for folder in candidate_dirs if (folder / worker_filename).is_file()), None)
if CODE_DIR is None:
    raise FileNotFoundError("Cannot locate the VECTOR education-analysis code folder. Check the notebook working directory.")
sys.path.insert(0, str(CODE_DIR)) if str(CODE_DIR) not in sys.path else None
import EDUCATION_20261001_romania_background_placement_worker as worker
worker = importlib.reload(worker)
print("Worker:", CODE_DIR / worker_filename)
print("Private source cache:", worker.PRIVATE)
print("Name-free county status:", worker.STATUS)
```

```python
# 3. Preview the queue. This cell never makes a network request.
preview_args = []
if COUNTIES:
    preview_args += ["--counties", *COUNTIES]
worker.main(preview_args)
```

```python
# 4. Acquire source pages. Live messages appear below while this cell runs.
# Change RUN_SOURCE_ACQUISITION to True in Cell 1, then rerun Cell 1 and this cell.
# Use Cursor's Interrupt button (or Control-C in a terminal) to stop. Each
# completed page is saved immediately; a request interrupted mid-download is
# simply retried on the next run. The county status may lag until its checkpoint.
if not RUN_SOURCE_ACQUISITION:
    print("No network request made. Set RUN_SOURCE_ACQUISITION = True in Cell 1, rerun Cell 1, then rerun this cell.")
else:
    run_args = ["--run", "--max-new-pages", str(MAX_NEW_PAGES),
                "--max-hours", str(MAX_HOURS),
                "--min-interval", str(MIN_INTERVAL_SECONDS),
                "--max-interval", str(MAX_INTERVAL_SECONDS)]
    if COUNTIES:
        run_args += ["--counties", *COUNTIES]
    if RETRY_PREVIOUS_HTTP_404:
        run_args.append("--retry-unresolved")
    print("Starting SOURCE-ONLY acquisition. Settings:", run_args, flush=True)
    worker.main(run_args)
```

```python
# 5. Read the name-free county status after a run or while a county is pending.
# During a county, the live text in Cell 4 is more current than this CSV.
import pandas as pd
from IPython.display import display

if worker.STATUS.exists(): 
    county_status = pd.read_csv(worker.STATUS)
    display(county_status.sort_values("county").reset_index(drop=True))
else:
    print("No county checkpoint yet. Run Cell 4 to begin source recovery.")
print("Status file:", worker.STATUS)
print("A placement_views_reconciled county still needs applicant-pool checks before research analysis.")
```

## How to read the status

- **`placement_views_reconciled`** means the targeted archived program reports match independent occupancy counts and exactly fill the county placement gap. It does **not** mean the national data are ready for analysis.
- **`unresolved`** covers pages not yet requested, missing archived pages, interrupted runs, count mismatches, or ambiguous signatures. No item is automatically labeled permanently unavailable.
- An HTTP 404 stays in a private unresolved-request ledger and is skipped on ordinary resumes. Charles can explicitly retry it or investigate another archived view. An HTTP 429 stops the run so the archive can recover.
- The notebook does not decide whether incomplete applicant rosters, unassigned reports, origin-gymnasium groups, or person linkage are acceptable. Charles must review those source gates before authorizing any national research product.
