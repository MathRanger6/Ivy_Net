```python
# CELL 3 — Preview what is already saved. This makes NO Wayback request.
# The expected starting point is 36 of 464 verified views, leaving 428.
worker.main(['--counties', *COUNTIES])
```

```python
# CELL 4 — Download missing gymnasium webpages, only if RUN_DOWNLOAD is True.
# You can interrupt this cell in Cursor. Every verified page is checkpointed immediately.
# Errors remain unresolved and can be retried on a later run; they are never called absent.
if not RUN_DOWNLOAD:
    print('Download is OFF. Set RUN_DOWNLOAD = True in Cell 1 and rerun Cell 1, then Cell 4.')
else:
    run_args = [
        '--run', '--counties', *COUNTIES,
        '--max-new-pages', str(MAX_NEW_PAGES),
        '--max-http-attempts', str(MAX_HTTP_ATTEMPTS),
        '--max-hours', str(MAX_HOURS),
        '--min-interval-seconds', str(MIN_INTERVAL_SECONDS),
        '--max-interval-seconds', str(MAX_INTERVAL_SECONDS),
        '--first-429-cooldown-seconds', str(FIRST_429_COOLDOWN_SECONDS),
        '--second-429-cooldown-seconds', str(SECOND_429_COOLDOWN_SECONDS),
        '--server-error-wait-seconds', str(SERVER_ERROR_WAIT_SECONDS),
    ]
    print('Starting source acquisition:', run_args, flush=True)
    worker.main(run_args)
```

```python
# CELL 5 — Read the name-free progress file after Cell 4 stops or you interrupt it.
# This is a local read only. No archived webpage is requested here.
import pandas as pd

if not worker.STATUS.exists():
    print('No progress file yet. Run Cell 4 with RUN_DOWNLOAD = True first.')
else:
    progress = pd.read_csv(worker.STATUS, dtype={'gymnasium_code': str}).fillna('')
    summary = progress.groupby(['county', 'status'], dropna=False).size().rename('gymnasiums').reset_index()
    display(summary)
    print('Verified:', int((progress['status'] == 'saved_verified').sum()), '/', len(progress))
    unresolved = progress[progress['status'].str.startswith('unresolved')]
    print('Unresolved school pages:', len(unresolved))
    if not unresolved.empty:
        display(unresolved[['county', 'gymnasium_code', 'status', 'last_http_status']])
    print('Progress file:', worker.STATUS)
```
