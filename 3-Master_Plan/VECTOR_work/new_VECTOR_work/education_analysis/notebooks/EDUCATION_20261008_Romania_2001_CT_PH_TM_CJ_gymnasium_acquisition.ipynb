# Romania 2001 — acquire gymnasium applicant reports for CT, PH, TM, and CJ

The October 8 offline audit found complete **school-result reports** and **incoming-candidate lists** for these four counties. The separate **gymnasium applicant reports** are still needed because they contain each participating applicant's examination and school-grade components. This notebook downloads those reports only. It does not match students, calculate peer pools, rank programs, or draw outcome plots.

**Run Cells 1–4 in order in Cursor.** Cell 3 previews the expected work without making an archive request. Cell 4 is the download. It prints the requested school, archive response, saved page, reason for any wait, and progress every 25 addresses. You can interrupt Cell 4; every verified page is saved immediately and reused on the next run. When it stops, run Cell 5 for a concise progress table. To resume, run Cell 4 again.

The private pages, which contain names and scores, stay under `~/Desktop/VECTOR_temp/romania_2001_four_county_school_source_pilot/school_reports/`. Only a name-free progress file is written to the research repository.


```python
# CELL 1 — Researcher-controlled settings. Edit these values in Cursor.
from pathlib import Path

# These four county codes come from the saved Ministry directory.
COUNTIES = ["CT", "PH", "TM", "CJ"]

# Cell 4 contacts the Wayback Machine only while this switch is True.
RUN_DOWNLOAD = True

# There are 791 listed gymnasium webpages across the four counties. These caps
# allow a complete run if the archive remains responsive, while still bounding it.
# A normal 302 redirect plus a 200 page response uses TWO HTTP attempts.
MAX_NEW_PAGES = 900
MAX_HTTP_ATTEMPTS = 3000
MAX_HOURS = 30.0

# The worker spaces requests adaptively inside these bounds. Its saved pacing
# and server backoff survive notebook restarts. A 429 triggers a 600-second
# cooldown, then 1500 seconds after a second consecutive 429; it stops after a
# third consecutive 429. HTTP 500/503 receives at least another 10-second wait.
MIN_INTERVAL_SECONDS = 21.0
MAX_INTERVAL_SECONDS = 50.0
FIRST_429_COOLDOWN_SECONDS = 600
SECOND_429_COOLDOWN_SECONDS = 1500
SERVER_ERROR_WAIT_SECONDS = 10.0

# This CSV contains county/school-code status and counts, never student records.
WORKSPACE = Path("/Users/charleslevine/Library/CloudStorage/Dropbox/1-Documents/00- Dissertation/0-Next_Chapter/Code_and_Data/New SQL and PY Code/Cursor Workspace PDE/3-Master_Plan/VECTOR_work/new_VECTOR_work")
STATUS_FILE = WORKSPACE / "education_analysis/outputs/romania_2001_four_new_county_source_reconciliation_20261008/gymnasium_acquisition_status.csv"

print("Counties:", ", ".join(COUNTIES))
print("Download enabled:", RUN_DOWNLOAD)
print("Request spacing:", MIN_INTERVAL_SECONDS, "to", MAX_INTERVAL_SECONDS, "seconds")
print("Name-free progress file:", STATUS_FILE)
```

```python
# CELL 2 — Load the existing checkpointing acquisition worker.
# If the Python worker is edited while this notebook is open, restart the kernel
# and rerun this cell so Python loads the updated version.
import importlib
import sys

CODE_DIR = WORKSPACE / "education_analysis/code"
if not CODE_DIR.is_dir():
    raise FileNotFoundError(f"Code folder not found: {CODE_DIR}")
if str(CODE_DIR) not in sys.path:
    sys.path.insert(0, str(CODE_DIR))
worker = importlib.import_module("EDUCATION_20261002_romania_three_county_gymnasium_acquisition")
worker = importlib.reload(worker)
print("Worker:", worker.__file__)
print("Private source folder:", worker.PAGES)
```

```python
# CELL 3 — Read-only preview. This makes NO Wayback request.
# It enumerates the saved Ministry school directories and checks for pages
# already saved in the private cache. Repeating it is safe.
worker.main(["--counties", *COUNTIES, "--status-file", str(STATUS_FILE)])
```

```python
# CELL 4 — Download and checkpoint missing gymnasium reports.
# Each success is saved before the next address is tried. Network failures
# remain unresolved; a failed request is never interpreted as zero applicants.
# Interrupt this cell in Cursor whenever needed, then rerun to resume.
if not RUN_DOWNLOAD:
    print("Download is OFF. Set RUN_DOWNLOAD = True in Cell 1, rerun Cell 1, then Cell 4.")
else:
    run_args = [
        "--run", "--counties", *COUNTIES,
        "--status-file", str(STATUS_FILE),
        "--max-new-pages", str(MAX_NEW_PAGES),
        "--max-http-attempts", str(MAX_HTTP_ATTEMPTS),
        "--max-hours", str(MAX_HOURS),
        "--min-interval-seconds", str(MIN_INTERVAL_SECONDS),
        "--max-interval-seconds", str(MAX_INTERVAL_SECONDS),
        "--first-429-cooldown-seconds", str(FIRST_429_COOLDOWN_SECONDS),
        "--second-429-cooldown-seconds", str(SECOND_429_COOLDOWN_SECONDS),
        "--server-error-wait-seconds", str(SERVER_ERROR_WAIT_SECONDS),
    ]
    print("Starting SOURCE-ONLY acquisition for", ", ".join(COUNTIES), flush=True)
    worker.main(run_args)
```

```python
# CELL 5 — Read the name-free status file after Cell 4 stops or is interrupted.
# This is local-only: no Wayback requests. The table distinguishes verified,
# unresolved, and not-yet-attempted school reports in each county.
import pandas as pd

if not STATUS_FILE.exists():
    print("No acquisition status yet. Cell 3 is only a preview; run Cell 4 first.")
else:
    progress = pd.read_csv(STATUS_FILE, dtype={"gymnasium_code": str}).fillna("")
    for county in COUNTIES:
        local = progress[progress["county"] == county]
        verified = int((local["status"] == "saved_verified").sum())
        unresolved = int(local["status"].str.startswith("unresolved").sum())
        print(f"{county}: {verified}/{len(local)} verified; "
              f"{unresolved} unresolved; {len(local) - verified - unresolved} not yet attempted")
    unresolved_rows = progress[progress["status"].str.startswith("unresolved")]
    if not unresolved_rows.empty:
        display(unresolved_rows[["county", "gymnasium_code", "status", "last_http_status"]])
    print("Status file:", STATUS_FILE)
```
