```python
import yfinance as yf

# Define the stock list
positions = [("CGNX",70.23,7),("NVDA",220.99,46),("SPCX",160.94,60),("STK",50.97,77),("FUN",64.26,2),("RCL",80.63,6)]
net_total = 0
for position,cost,holding in positions:
    # Fetch the ticker data
    ticker = yf.Ticker(position)

    # Get the current price from the info dictionary
    current_price = ticker.info.get('currentPrice')

    # Calculate net gain/loss
    net_gain_loss = (current_price - cost) * holding
    
    # Add to net_total
    net_total += net_gain_loss

    print(f"The current price for {position} is: ${current_price}, the cost was ${cost}")
    print(f"       While holding {holding} shares, the net {"gain" if current_price > cost else "loss"} is ${net_gain_loss:,.2f}")
print(f"\nThe current net total {"gain" if net_total > 0 else "loss"} is ${net_total:,.2f}\n")
```
